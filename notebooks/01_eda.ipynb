```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import warnings
warnings.filterwarnings('ignore')

# Display settings
pd.set_option('display.max_columns', 100)
pd.set_option('display.float_format', '{:.2f}'.format)

# Load data
df = pd.read_csv('../data/raw/application_train.csv')

print(f"Shape: {df.shape}")
print(f"Rows: {df.shape[0]:,}")
print(f"Columns: {df.shape[1]}")
```

```python
# First 5 rows
df.head()
```

```python
# Data types and non-null counts
df.info()
```

```python
# Basic statistics
df.describe().T
```

```python
target_counts = df['TARGET'].value_counts()
target_pct = df['TARGET'].value_counts(normalize=True) * 100

print("Target Distribution:")
print(f"  Repaid (0): {target_counts[0]:,} ({target_pct[0]:.1f}%)")
print(f"  Defaulted (1): {target_counts[1]:,} ({target_pct[1]:.1f}%)")
```

```python
# Visualize
fig = px.pie(
    values=target_counts.values,
    names=['Repaid (0)', 'Defaulted (1)'],
    title='Loan Default Distribution',
    color_discrete_sequence=['#2ecc71', '#e74c3c']
)
fig.show()
```

```python
# Calculate missing percentages
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({
    'Missing Count': missing,
    'Missing %': missing_pct
}).sort_values('Missing %', ascending=False)

missing_df = missing_df[missing_df['Missing Count'] > 0]
print(f"Columns with missing values: {len(missing_df)}")
missing_df.head(20)
```

```python
# Visualize top 30 columns with most missing data
top_missing = missing_df.head(30)

fig = px.bar(
    top_missing,
    x=top_missing.index,
    y='Missing %',
    title='Top 30 Columns by Missing Data %',
    color='Missing %',
    color_continuous_scale='Reds'
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```

```python
# Numerical features distribution
numerical_cols = ['AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 
                  'AMT_GOODS_PRICE', 'DAYS_BIRTH', 'DAYS_EMPLOYED']

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    axes[i].hist(df[col].dropna(), bins=50, color='steelblue', edgecolor='white')
    axes[i].set_title(col)
    axes[i].set_xlabel('Value')
    axes[i].set_ylabel('Count')

plt.suptitle('Key Financial Features Distribution', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Default rate by income type
default_by_income = df.groupby('NAME_INCOME_TYPE')['TARGET'].mean().sort_values(ascending=False)

fig = px.bar(
    x=default_by_income.index,
    y=default_by_income.values,
    title='Default Rate by Income Type',
    labels={'x': 'Income Type', 'y': 'Default Rate'},
    color=default_by_income.values,
    color_continuous_scale='Reds'
)
fig.show()
```

```python
# Default rate by education
default_by_edu = df.groupby('NAME_EDUCATION_TYPE')['TARGET'].mean().sort_values(ascending=False)

fig = px.bar(
    x=default_by_edu.index,
    y=default_by_edu.values,
    title='Default Rate by Education Level',
    labels={'x': 'Education', 'y': 'Default Rate'},
    color=default_by_edu.values,
    color_continuous_scale='Oranges'
)
fig.show()
```

```python
# Correlation of numerical features with TARGET
corr_with_target = df.select_dtypes(include=[np.number]).corr()['TARGET'].drop('TARGET')
corr_sorted = corr_with_target.abs().sort_values(ascending=False).head(20)

fig = px.bar(
    x=corr_sorted.index,
    y=corr_sorted.values,
    title='Top 20 Features Correlated with Default',
    labels={'x': 'Feature', 'y': 'Absolute Correlation'},
    color=corr_sorted.values,
    color_continuous_scale='Blues'
)
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```
